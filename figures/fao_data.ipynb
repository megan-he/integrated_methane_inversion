```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
africa = pd.read_csv('./data/FAOSTAT_cattle_Africa.csv')
# keep only year column and Value column
africa = africa[['Year', 'Value']]
africa.columns = ['Year', 'Cattle']
africa.head()

# set year as index
africa = africa.set_index('Year')

# plot
fig, ax = plt.subplots(figsize=(7,5))
plt.rcParams.update({'font.size': 16})
plt.plot(africa.index, africa['Cattle'], linewidth=4, marker='o', markersize=10)
plt.ylim(3.5e8, 4e8)
plt.xlabel('Year')
plt.ylabel('Cattle')
plt.show()
```
